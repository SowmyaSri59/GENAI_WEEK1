Task - 1 ( Python Basics & Operator)


```python
# Accept two integers from the user
a = int(input("Enter a number:"))
b = int(input("Enter a number:"))

# arithmetic operations
add = a + b
sub = a - b
mul = a * b
div = a / b
modulo = a % b
power = a ** b

# operators
greater = (a > b)
smaller = (a < b)
equal = (a == b)
not_equal = (a != b)
ope1 = (a > 5) and (b < 5)
ope2 = (a < 7) or (b > 10)
ope3 = not(a == 100)

# using formatted string
print(f"\n arithmetic operations")
print(f"Addition:{a} + {b} = {add}")
print(f"subtraction:{a} - {b} = {sub}")
print(f"Multiplication:{a} * {b} = {mul}")
print(f"Division:{a} / {b} = {div}")
print(f"Modulo Division:{a} % {b} = {modulo}")
print(f"Power:{a} ** {b} = {power}")

print(f"\n Operators")
print(f"{a} > {b}: {greater}")
print(f"{a} < {b}: {smaller}")
print(f"{a} == {b}: {equal}")
print(f"{a} != {b}: {not_equal}")
print(f"Both positive (AND): {ope1}")
print(f"Any greater than 100 (OR): {ope2}")
print(f"Not equal (NOT): {ope3}")
```

Task - 2 Lists and Arrays

```python
# 1) list of 10 random integers
import random
import numpy as np
num = [random.randint(1, 100) for i in range(10)]
print(num)

# 2) Perform operations
  # a) Add and Remove
num.append(75)
print(num)
num.remove(75)
print(num)
  # b) Max and Min
Max =max(num)
print(Max)
Min = min(num)
print(Min)
  # c) Sort
sort = sorted(num)
print(sort)

# 3) List to Numpy array
  # Mean, median, std
num_array = np.array(num)
Mean = np.mean(num_array)
Median= np.median(num_array)
sd = np.std(num_array)
print(Mean)
print(Median)
print(sd)


```

Task - 3 Dictonaries and Sets

```python
# 1) Create dictinary
student = {'name' : 'Priya', 'course' : 'Data Science', 'marks' : 93}

# 2) Add grade
marks = student['marks']
if marks >= 90:
    grade = 'A'
elif marks >= 75:
    grade = 'B'
elif marks >= 60:
    grade = 'C'
else:
    grade = 'D'

student['grade'] = grade

# 3) print("--- Student Information ---")
for key, value in student.items():
    print(f"{key}: {value}")

# 4) Create two sets of AI tools
ai_tools_set1 = {'ChatGPT', 'Claude', 'Gemini', 'Copilot', 'Midjourney'}
ai_tools_set2 = {'ChatGPT', 'DALL-E', 'Stable Diffusion', 'Gemini', 'Bard'}

print(f"\n--- AI Tools Sets ---")
print(f"Set 1: {ai_tools_set1}")
print(f"Set 2: {ai_tools_set2}")
print(f"Union: {ai_tools_set1.union(ai_tools_set2)}")
print(f"Intersection: {ai_tools_set1.intersection(ai_tools_set2)}")
print(f"Difference (Set1 - Set2): {ai_tools_set1.difference(ai_tools_set2)}")

```

Task - 4 File Handling


```python
# create ai_students.txt & write details
students_data = """Priya,93,A
Sri,78,B
Ashini,70,C
Gayathri,87,B
Ramya,58,D"""

with open("ai_students.txt", "w") as file:
  file.write(students_data)
  print("file created successfully")

# Read and Display
with open("ai_students.txt","r") as file:
  for line in file:
        name, marks, grade = line.strip().split(',')
        if int(marks) > 75:
            print(f"Name: {name}, Marks: {marks}, Grade: {grade}")
```

```python
from datetime import datetime
# 1) Input from user
user_prompt = input("Enter your AI prompt: ")

# 2 & 3) Save prompt with timestamp (append mode)
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
log_entry = f"[{timestamp}] {user_prompt}\n"

with open('prompt_history.txt', 'a') as file:
    file.write(log_entry)

print(f"\n Prompt saved successfully at {timestamp}")

# Display all saved prompts
print("\n  Prompt History")
with open('prompt_history.txt', 'r') as file:
    print(file.read())
```
